```python
from sklearn.linear_model import LinearRegression 
```

```python
import pandas as pd
```

```python
df = pd.read_csv("C:/Users/user/Downloads/financial_regression.csv")
```

```python
df.head(5)
```

```python
df.info()
```

```python
df = df.drop(df.columns[0:43], axis=1)
```

```python
df = df.dropna()
```

```python
df.head(10)
```

```python
y=df[["gold close"]]
x=df[["gold low","gold high"]]
```

```python
l = LinearRegression()
model = l.fit(x,y) #y=ax+b 
```

```python
model.predict([[110.79,112.37]]) #input verilir output alınır
```

```python
model.score(x,y)
```

```python
model.coef_ #katsayı
```

```python
model.intercept_ #sabit değişken
```
